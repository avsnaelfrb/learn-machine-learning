# 04 - Preprocessing, Scaling, dan Pipeline

Banyak model machine learning sensitif terhadap skala angka.

Contoh:
- fitur A bernilai 0 sampai 10
- fitur B bernilai 0 sampai 100000

Model seperti KNN bisa terlalu terpengaruh oleh fitur yang skalanya besar.

## Step 1 - Import Library

````python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
````

## Step 2 - Load Dataset

````python
wine = load_wine()

X = wine.data
y = wine.target

print("Shape X:", X.shape)
print("Shape y:", y.shape)
print("Nama fitur pertama sampai kelima:", wine.feature_names[:5])
print("Nama kelas:", wine.target_names)
````

## Step 3 - Split Data Sebelum Scaling

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Train:", X_train.shape)
print("Test :", X_test.shape)
````

Aturan penting:

Split data dulu, baru lakukan preprocessing.

Tujuannya agar informasi dari data test tidak bocor ke proses training.

## Step 4 - Model Tanpa Scaling

````python
model_tanpa_scaling = KNeighborsClassifier(n_neighbors=5)
model_tanpa_scaling.fit(X_train, y_train)

pred_tanpa_scaling = model_tanpa_scaling.predict(X_test)
acc_tanpa_scaling = accuracy_score(y_test, pred_tanpa_scaling)

print("Accuracy tanpa scaling:", acc_tanpa_scaling)
````

## Step 5 - Model Dengan Scaling Manual

````python
scaler = StandardScaler()

# fit hanya pada data train
scaler.fit(X_train)

# transform train dan test memakai scaler yang sama
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

model_dengan_scaling = KNeighborsClassifier(n_neighbors=5)
model_dengan_scaling.fit(X_train_scaled, y_train)

pred_dengan_scaling = model_dengan_scaling.predict(X_test_scaled)
acc_dengan_scaling = accuracy_score(y_test, pred_dengan_scaling)

print("Accuracy dengan scaling:", acc_dengan_scaling)
````

## Step 6 - Pipeline: Cara Lebih Rapi

````python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier(n_neighbors=5)),
])

pipeline.fit(X_train, y_train)
pred_pipeline = pipeline.predict(X_test)
acc_pipeline = accuracy_score(y_test, pred_pipeline)

print("Accuracy pipeline:", acc_pipeline)
````

Pipeline membuat preprocessing dan model menjadi satu rangkaian.

Keuntungannya:
- kode lebih rapi
- risiko lupa scaling data baru lebih kecil
- cocok untuk eksperimen model yang lebih serius

## Step 7 - Prediksi Data Baru Dengan Pipeline

````python
data_baru = [X_test[0]]
prediksi = pipeline.predict(data_baru)

print("Target asli   :", wine.target_names[y_test[0]])
print("Prediksi model:", wine.target_names[prediksi[0]])
````
