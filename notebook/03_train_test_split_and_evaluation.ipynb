# 03 - Train/Test Split dan Evaluasi Model

Notebook ini fokus ke pertanyaan penting:

**Bagaimana kita tahu model benar-benar belajar, bukan hanya menghafal?**

Kita akan membandingkan performa model pada data train dan data test.

## Step 1 - Import Library

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

## Step 2 - Load Dataset

```python
data = load_breast_cancer()

X = data.data
y = data.target

print("Shape X:", X.shape)
print("Shape y:", y.shape)
print("Kelas:", data.target_names)
```

Dataset ini dipakai hanya untuk latihan teknis machine learning.

Targetnya adalah dua kelas: `malignant` dan `benign`.

## Step 3 - Split Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Jumlah train:", len(X_train))
print("Jumlah test :", len(X_test))
```

## Step 4 - Buat Fungsi Evaluasi Kecil

```python
def train_dan_evaluasi(jumlah_tetangga):
    model = KNeighborsClassifier(n_neighbors=jumlah_tetangga)
    model.fit(X_train, y_train)

    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)

    acc_train = accuracy_score(y_train, pred_train)
    acc_test = accuracy_score(y_test, pred_test)

    return acc_train, acc_test
```

## Step 5 - Bandingkan Beberapa Model

```python
daftar_k = [1, 3, 5, 9, 15, 25]

for k in daftar_k:
    acc_train, acc_test = train_dan_evaluasi(k)
    gap = acc_train - acc_test
    print(f"k={k:>2} | train={acc_train:.3f} | test={acc_test:.3f} | gap={gap:.3f}")
```

Cara membaca hasil:
- Train tinggi, test jauh lebih rendah: model cenderung menghafal data train
- Train dan test sama-sama rendah: model belum cukup menangkap pola
- Train dan test sama-sama cukup tinggi: model lebih masuk akal

## Step 6 - Pilih Model Final

```python
model_final = KNeighborsClassifier(n_neighbors=5)
model_final.fit(X_train, y_train)

y_pred = model_final.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy model final:", accuracy)
```

## Step 7 - Cek Beberapa Prediksi

```python
for i in range(10):
    aktual = data.target_names[y_test[i]]
    prediksi = data.target_names[y_pred[i]]
    benar = aktual == prediksi
    print(f"Data ke-{i} | Aktual: {aktual:>9} | Prediksi: {prediksi:>9} | Benar: {benar}")
```

## Latihan Praktikal

Coba ubah:
- `test_size` menjadi 0.2, 0.3, dan 0.4
- `random_state` menjadi angka lain
- `n_neighbors` untuk model final

Amati apakah accuracy stabil atau mudah berubah.
